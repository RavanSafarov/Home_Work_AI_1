```python
my_columns = ['Имя','Проект','Часы']
work_data = [
    ["Расул", "Сайт", 40],
    ["Расул", "Мобильное приложение", 25],
    ["Лейла", "Сайт", 35],
    ["Лейла", "CRM", 30],
    ["Эльчин", "Мобильное приложение", 50],
    ["Эльчин", "CRM", 20],
    ["Анна", "Сайт", 28],
    ["Анна", "Мобильное приложение", 32],
    ["Игорь", "CRM", 45],
    ["Игорь", "Сайт", 15]
]
df_work = pd.DataFrame(work_data, columns=my_columns)
df_work
```
